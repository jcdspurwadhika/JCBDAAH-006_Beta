## 1.1 Hotel Industry Overview

Industri perhotelan merupakan salah satu sektor jasa yang menyediakan layanan akomodasi untuk berbagai kebutuhan perjalanan, baik bisnis maupun rekreasi. Dalam operasionalnya, hotel memiliki kapasitas kamar yang terbatas dan kamar yang tidak terjual pada suatu periode tidak dapat disimpan untuk dijual kembali di periode berikutnya. Karena itu, pengelolaan jumlah kamar dan permintaan menjadi salah satu bagian penting dalam bisnis hotel.

Hotel juga perlu memahami pola reservasi pelanggan, mulai dari waktu melakukan booking, karakteristik tamu, sampai perubahan status reservasi. Jumlah booking yang tinggi belum tentu menghasilkan tingkat occupancy yang sama apabila terdapat booking yang berubah atau dibatalkan sebelum tanggal kedatangan.

Disini reservation management menjadi cukup penting karena berkaitan langsung dengan occupancy planning dan pengelolaan revenue hotel. Dengan memahami pola booking dan cancellation, pihak hotel dapat mengetahui karakteristik reservasi yang perlu mendapat perhatian lebih dan menggunakan informasi tersebut sebagai bahan dalam menentukan strategi pengelolaan reservasi.

## 1.2 Background

Berdasarkan dataset reservasi hotel yang digunakan pada project ini, terdapat booking yang pada akhirnya dibatalkan oleh customer. Hal ini menyebabkan jumlah booking yang tercatat belum tentu sama dengan jumlah reservasi yang benar-benar terealisasi menjadi actual stay.

Setelah proses data cleaning, terdapat **87.395 booking**, dimana **24.025 booking atau 27,49% berstatus canceled**. Persentase tersebut menunjukan bahwa cancellation menjadi bagian yang cukup besar dari keseluruhan reservasi dan dapat memengaruhi perencanaan occupancy serta potential room value yang sebelumnya diharapkan dari booking tersebut.

Cancellation juga tidak selalu memiliki pola yang sama pada setiap reservasi. Beberapa karakteristik seperti hotel type, lead time, market segment, deposit type, dan karakteristik booking lainnya dapat memiliki cancellation rate yang berbeda-beda.

Karena itu, analisis ini dilakukan untuk melihat pola cancellation secara lebih detail, mengetahui kelompok booking mana yang memiliki cancellation rate lebih tinggi, serta mengestimasi potential room value dari booking yang dibatalkan. Hasil analisis digunakan untuk menentukan kelompok booking yang perlu diprioritaskan dan menyusun rekomendasi terkait cancellation.

#### 2.10.5.1 Identifikasi Outlier pada ADR

Mengidentifikasi nilai ekstrem pada variabel `adr` menggunakan metode Interquartile Range (IQR) untuk mengetahui jumlah observasi yang berpotensi menjadi outlier secara statistik.

```python
Q1 = df_hotel['adr'].quantile(0.25)
Q3 = df_hotel['adr'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print('Q1:', Q1)
print('Q3:', Q3)
print('IQR:', IQR)
print('Lower Bound:', lower_bound)
print('Upper Bound:', upper_bound)
```

```python
# Cari total kandidat outlier
adr_outliers = df_hotel[
    (df_hotel['adr'] < lower_bound) |
    (df_hotel['adr'] > upper_bound)
]

print('Jumlah kandidat outlier ADR:', len(adr_outliers))
```

```python
# Memisahkan jenis outlier
print('ADR di bawah lower bound:', (df_hotel['adr'] < lower_bound).sum())
print('ADR di atas upper bound:', (df_hotel['adr'] > upper_bound).sum())
```

**Kesimpulan:** Berdasarkan hasil identifikasi menggunakan metode IQR di atas, terdapat sekiatr 3.793 observasi (3,18%) yang teridentifikasi sebagai kandidat outlier pada variabel adr. Seluruh kandidat berada di atas upper bound sebesar 211,065. Namun, kandidat outlier tersebut belum langsung dikategorikan sebagai anomaly atau dihapus, karena nilai ADR yang tinggi masih dapat merepresentasikan kondisi reservasi yang valid. Oleh karena itu, diperlukan pemeriksaan lebih lanjut berdasarkan karakteristik reservasi sebelum menentukan treatment.

Pada project ini, feature engineering mendukung analisis:
- karakteristik reservasi,
- lama menginap,

## 2.12.1 Total Stay Duration

### Konsep

Dataset menyediakan lama menginap dalam dua variabel:
- `stays_in_weekend_nights`
- `stays_in_week_nights`

Kedua variabel tersebut digabungkan untuk memperoleh total lama menginap pelanggan.

### Formula

`Total Stay Duration = Weekend Nights + Week Nights`

### Business Reasoning

Total lama menginap memberikan gambaran yang lebih sederhana mengenai karakteristik reservasi.

Variabel ini juga digunakan untuk menghitung estimasi nilai reservasi.


```python
df_hotel['total_stay_duration'] = (
    df_hotel['stays_in_weekend_nights'] +
    df_hotel['stays_in_week_nights']
)

df_hotel[['stays_in_weekend_nights',
          'stays_in_week_nights',
          'total_stay_duration']].head()
```

## 2.12.2 Total Guests

### Konsep

Jumlah tamu awalnya tersebar dalam beberapa variabel:
- `adults`
- `children`
- `babies`

Variabel tersebut digabungkan menjadi jumlah total tamu dalam satu reservasi.

### Formula

`Total Guests = Adults + Children + Babies`

### Business Reasoning

Total guests membantu menggambarkan ukuran reservasi dan karakteristik pelanggan.


```python
df_hotel['total_guests'] = (
    df_hotel['adults'].fillna(0) +
    df_hotel['children'].fillna(0) +
    df_hotel['babies'].fillna(0)
)

df_hotel[['adults',
          'children',
          'babies',
          'total_guests']].head()
```

## 2.12.3 Potential Room Value

### Konsep

Dataset tidak menyediakan nilai total reservasi secara langsung.

Untuk memperkirakan nilai reservasi yang terdampak cancellation, dibuat variabel turunan menggunakan ADR dan total lama menginap.

### Formula

`Potential Room Value = ADR × Total Stay Duration`

### Catatan

Potential Room Value merupakan proxy nilai reservasi yang terdampak dan bukan actual revenue loss karena belum mempertimbangkan:
- biaya operasional,
- kemungkinan kamar dijual kembali,
- perubahan harga kamar.


```python
df_hotel['potential_room_value'] = (
    df_hotel['adr'] *
    df_hotel['total_stay_duration']
)

df_hotel[['adr',
          'total_stay_duration',
          'potential_room_value']].head()
```

## 3.7 Objective 2 — Statistical Validation

Untuk variabel kategorikal, digunakan **Chi-Square Test of Independence** dengan **Cramer's V** sebagai ukuran kekuatan association. Untuk variabel numerik, digunakan **Mann-Whitney U**.


```python
from scipy.stats import chi2_contingency, mannwhitneyu

cat_cols = ['hotel','market_segment','deposit_type']
chi_rows=[]
for col in cat_cols:
    table = pd.crosstab(df_hotel[col], df_hotel['is_canceled'])
    chi2,p,dof,expected = chi2_contingency(table)
    n=table.values.sum(); r,k=table.shape
    phi2=chi2/n
    v=(phi2/min(k-1,r-1))**0.5 if min(k-1,r-1)>0 else np.nan
    chi_rows.append({'variable':col,'chi2':chi2,'p_value':p,'cramers_v':v})
chi_results=pd.DataFrame(chi_rows)
display(chi_results.round(4))
```

```python
mw_rows=[]
for col in ['lead_time','adr']:
    a=df_hotel.loc[df_hotel.is_canceled==0,col].dropna()
    b=df_hotel.loc[df_hotel.is_canceled==1,col].dropna()
    u,p=mannwhitneyu(a,b,alternative='two-sided')
    mw_rows.append({'variable':col,'median_not_canceled':a.median(),'median_canceled':b.median(),'U_statistic':u,'p_value':p})
mw_results=pd.DataFrame(mw_rows)
display(mw_results.round(4))
```

**Insight statistik:** Hotel type, market segment, dan deposit type menunjukkan evidence adanya association dengan `is_canceled` berdasarkan Chi-Square Test (**p-value < 0,001**). `lead_time` dan `adr` juga menunjukkan perbedaan distribusi antara canceled dan not canceled berdasarkan Mann-Whitney U (**p-value < 0,001**).

Karena jumlah observasi besar, p-value hampir pasti sangat kecil, sehingga yang lebih informatif adalah **effect size**:

| Variabel | Ukuran | Nilai | Interpretasi |
|---|---|---:|---|
| `market_segment` | Cramér's V | **0,2210** | Asosiasi terkuat di antara variabel kategorikal (kecil–sedang) |
| `deposit_type` | Cramér's V | 0,1651 | Asosiasi kecil–sedang, perlu dibaca bersama temuan anomali Non Refund |
| `hotel` | Cramér's V | 0,0716 | Asosiasi sangat lemah |
| `lead_time` | Median canceled vs not canceled | **80** vs **38** hari | Booking canceled dibuat jauh lebih awal |
| `adr` | Median canceled vs not canceled | **109,8** vs **94,5** (**+16,2%**) | Booking canceled cenderung bertarif lebih tinggi |

**Business insight:**
- Urutan kekuatan sinyal: **market segment** dan **lead time** menjadi dasar utama segmentasi risiko, deposit type menjadi area investigasi, dan hotel type dipakai sebagai pembeda level benchmark.
- ADR booking canceled yang lebih tinggi berarti **exposure per booking juga lebih besar** (dibahas di 3.8).
- Hasil statistik digunakan untuk memperkuat temuan descriptive analysis, **bukan untuk menyimpulkan causal relationship**. Seluruh uji bersifat univariat dan belum mengontrol confounder (misalnya Online TA yang mungkin berkorelasi dengan lead time panjang).


## 3.8 Objective 3 — Potential Room Value dari Cancelled Booking

**Formula:** `Potential Room Value = ADR × Total Nights`

Nilai ini merupakan **proxy potential value**, bukan actual lost revenue karena belum memperhitungkan resale, discount, tax, atau realized revenue.


```python
df_hotel['total_nights'] = df_hotel['stays_in_weekend_nights'] + df_hotel['stays_in_week_nights']
df_hotel['potential_room_value'] = df_hotel['adr'] * df_hotel['total_nights']

cancelled_value = df_hotel.loc[df_hotel.is_canceled==1,'potential_room_value'].sum()
overall_value = df_hotel['potential_room_value'].sum()
value_table = pd.DataFrame({
    'Metric':[
        'Potential room value - all reservations',
        'Potential room value - cancelled reservations',
        'Share of potential room value from cancelled reservations'
    ],
    'Value':[overall_value,cancelled_value,cancelled_value/overall_value*100]
})
display(value_table)
```

```python
pvr_hotel = df_hotel[df_hotel.is_canceled==1].groupby('hotel').agg(
    cancelled_booking=('is_canceled','size'),
    potential_room_value=('potential_room_value','sum')
)
pvr_hotel['share_of_cancelled_value_pct'] = pvr_hotel['potential_room_value']/cancelled_value*100
display(pvr_hotel.round(2))
```

```python
plt.figure(figsize=(8,5))
plot=pvr_hotel.reset_index()
plt.bar(plot['hotel'],plot['potential_room_value'])
plt.title('Potential Room Value of Cancelled Bookings by Hotel')
plt.xlabel('Hotel Type')
plt.ylabel('Potential Room Value')
plt.tight_layout()
plt.show()
```

```python
pvr_lead = df_hotel[df_hotel.is_canceled==1].groupby('lead_time_group',observed=False).agg(
    cancelled_booking=('is_canceled','size'),
    potential_room_value=('potential_room_value','sum')
)
pvr_lead['share_of_cancelled_value_pct']=pvr_lead['potential_room_value']/cancelled_value*100
display(pvr_lead.round(2))
```

```python
plt.figure(figsize=(10,5))
plot=pvr_lead.reset_index()
plt.bar(plot['lead_time_group'].astype(str),plot['potential_room_value'])
plt.title('Potential Room Value of Cancelled Bookings by Lead Time Group')
plt.xlabel('Lead Time Group (days)')
plt.ylabel('Potential Room Value')
plt.tight_layout()
plt.show()
```

**Insight:**
- Potential room value dari cancelled booking sekitar **11,48 juta**, atau **33,32%** dari total **34,46 juta**. Porsi ini **lebih besar daripada porsi cancelled booking (27,49%)**, artinya cancellation condong terjadi pada booking bernilai lebih tinggi. Rata-rata potential room value per cancelled booking ≈ **478,0** vs ≈ **362,6** pada not canceled (**+31,8%**).
- **Per hotel:** City Hotel menyumbang **6,61 juta (57,6%)** dari potential value cancelled, padahal menyumbang 66,8% cancelled booking. Resort Hotel menyumbang **4,87 juta (42,4%)** dari 33,2% cancelled booking. Rata-rata nilai per cancelled booking **Resort ≈ 610,5** vs **City ≈ 412,1** (**+48%**).
- **Per lead time:** booking >30 hari menyumbang **81,26%** potential value cancelled (≈9,33 juta). Kelompok **91–180 hari (30,15%, 3,46 juta)** dan **31–90 hari (28,83%, 3,31 juta)** terbesar. Rata-rata nilai per cancelled booking naik seiring lead time (≈250 pada 0–7 hari → ≈543 pada 91–180 → ≈564 pada 181–365). Kelompok 366+ hanya **0,46%** dari value (≈227 per booking).

**Business insight:**
- Lead time panjang berarti **probabilitas cancel tinggi sekaligus nilai per booking tinggi**. Kelompok **91–365 hari** (36,77% rate, 45,06% dari cancellation, **51,97% dari potential value**) adalah area dengan exposure finansial terbesar.

## 6.1 Strategy 1 — Proactive Confirmation untuk Long-Lead Booking (P1)

### Finding
- Cancellation rate meningkat dari **8,43%** (0–7 hari) menjadi **40,88%** (366+ hari). Booking >30 hari memiliki rate **34,76%** vs **16,42%** pada ≤30 hari.

### Justification
Long-lead booking memenuhi empat kriteria sekaligus: **rate tinggi, volume besar, nilai per booking tinggi, dan waktu untuk bertindak masih cukup**. Selain itu, lead time sudah diketahui saat booking dibuat sehingga proses dapat dijalankan tanpa data tambahan.

1. Booking dibuat → sistem menandai tier berdasarkan lead time.
2. Kirim reminder/reconfirmation sesuai tier, dengan cara konfirmasi yang mudah (satu klik atau balas pesan).
3. Jika customer membatalkan atau tidak merespons, tim reservasi menindaklanjuti dan segera memperbarui inventory agar kamar dapat dijual kembali.
4. Cancellation yang diketahui lebih awal dilaporkan ke Revenue Manager sebagai input forecast.

| Pengetatan policy menurunkan konversi Online TA | Uji pada subset, jadikan jumlah stay terealisasi sebagai guardrail |
| Segmentasi hanya berdasar variabel univariat (confounding) | Analisis lanjutan multivariat (misalnya logistic regression) pada Future Work |

Pendekatan ini menjaga agar rekomendasi tetap berbasis pada pola yang ditemukan dalam dataset dan tidak menganggap association sebagai causal effect.


# BAB 7 CONCLUSION

## 7.1 Analysis Result

Pola cancellation berdasarkan hotel type dan lead time** | Cancellation naik konsisten seiring lead time di kedua hotel, dan City Hotel lebih tinggi di setiap kelompok lead time | City **30,04%** vs Resort **23,48%**; lead time **8,43% → 40,88%**; >30 hari = 60,36% booking tetapi 76,32% cancellation; City >30 hari = 50,27% cancellation |

## 7.2 Main Conclusion

Berdasarkan **87.395 booking** setelah cleaning, terdapat **24.025 cancelled booking** sehingga baseline **Cancellation Rate sebesar 27,49%**. Tingkat ini menunjukkan ketidakpastian booking yang relevan bagi perencanaan occupancy dan room availability.

1. **Kapan risiko tinggi: lead time panjang.** Cancellation rate meningkat dari **8,43%** (0–7 hari) menjadi **40,88%** (366+ hari). Booking >30 hari menyumbang **76,32%** cancellation dan **81,26%** cancelled potential value, sehingga menjadi titik kontrol utama. Kelompok 366+ memiliki rate tertinggi tetapi volumenya kecil, sehingga cukup dijadikan *watchlist*.
2. **Di mana risiko terkonsentrasi: Online TA dan City Hotel.** Online TA menyumbang **75,94%** cancellation (rate 35,35%) namun juga **52,66%** stay terealisasi. City Hotel (30,04%) lebih tinggi daripada Resort Hotel (23,48%) dan City >30 hari saja menyumbang separuh seluruh cancellation. Hotel type berfungsi sebagai pembeda level risiko (asosiasi lemah), sedangkan market segment dan lead time adalah sinyal yang lebih kuat.
3. **Seberapa besar dampaknya: exposure finansial terkonsentrasi di long-lead.** Cancelled booking mewakili **≈11,48 juta (33,32%)** potential room value, dengan nilai per booking lebih tinggi dibanding not canceled. Kelompok 91–365 hari memiliki exposure terbesar (**51,97%**). Nilai ini adalah **proxy exposure**, bukan actual lost revenue.
4. **Deposit type memerlukan investigasi, bukan kesimpulan.** Hampir seluruh cancellation berasal dari booking No Deposit (95,80%), tetapi rate Non Refund yang sangat tinggi (94,70%) bertolak belakang dengan ekspektasi dan perlu diaudit sebelum dipakai untuk keputusan kebijakan.

## 7.3 Objective Analysis

| Objective | Analysis Insight |
|---|---|
| 1. Pola berdasarkan hotel type dan lead time | Kelompok di atas baseline teridentifikasi: City Hotel dan booking >30 hari |
| 2. Karakteristik berdasarkan market segment dan deposit type | Online TA (volume dan rate tinggi); deposit type perlu audit |
| 3. Estimasi potential room value | ≈11,48 juta; kontribusi terbesar dari lead time 31–180 hari dan City Hotel |
| 4. Rekomendasi minimal 3 potential action | 4 strategi dengan target, KPI, dan trade-off |

**Kelayakan target 2 pp:** kemungkinan target dapat tercapai jika sekitar **9,5% cancellation booking >30 hari** berhasil dicegah, atau jika rate Online TA turun **≈3,39 pp**. Skala usaha ini terlihat realistis, tetapi efektivitas intervensi belum terbukti karena analisis hanya menggunakan dataset yang terbatas. Angka **1.748 booking** dan **835.503 potential value** adalah **skenario ilustratif**, bukan jaminan hasil.

## 7.4 Priority Recommnedation & Future Works

1. **Mulai pilot proactive confirmation** untuk booking >30 hari (fokus 91–365 hari) di City Hotel, dengan kelompok kontrol dan KPI yang disepakati.
2. **Lakukan drill-down Online TA** per agent, country, dan lead time untuk menemukan sumber konsentrasi cancellation sebelum mengubah kebijakan channel.
3. **Terapkan tiering risiko × potential room value** agar effort reconfirmation sebanding dengan exposure.
4. **Audit data deposit type (Non Refund)** sebelum merancang pilot deposit/cancellation policy.
5. Pantau **Cancellation Rate** sebagai KPI utama, dengan cancelled volume, potential room value exposure, dan *early cancellation notice* sebagai supporting metrics yang dapat disajikan dalam bentuk Dashboard untuk Stakeholder yang telah ditentukan (Design Dashboard pada Bab 9).

## 7.5 INterpretaiton

Seluruh temuan menunjukkan **asosiasi, bukan causal relationship**. Potential room value adalah **proxy** yang belum memperhitungkan resale, diskon, pajak, atau biaya. Uji statistik bersifat univariat dan belum mengontrol confounder, sehingga analisis lanjutan (multivariat dan pengujian efektivitas intervensi) menjadi bagian dari Future Work.


# BAB 8 LIMITATIONS & FUTURE WORK

## 8.1 Limitations

- Dataset hanya merepresentasikan booking yang tersedia pada dataset dan tidak memberikan konteks operasional hotel secara langsung.
- Analisis statistik menunjukkan association/difference, bukan causal relationship.
- **Potential Room Value** merupakan proxy dan bukan actual realized revenue atau confirmed revenue loss.
- Kelompok dengan volume sangat kecil, seperti 366+ hari pada Resort Hotel, perlu diinterpretasikan dengan hati-hati.
- Analisis belum memiliki informasi seperti alasan cancellation, biaya komunikasi, customer response terhadap reminder, resale rate, atau actual room revenue setelah cancellation.

## 8.2 Future Work

- Tambahkan data actual revenue dan room resale setelah cancellation untuk mengestimasi financial impact secara lebih akurat.
- Tambahkan cancellation reason untuk membedakan cancellation yang dapat dan tidak dapat dipengaruhi oleh intervention.
- Uji efektivitas proactive confirmation melalui controlled pilot/A-B test jika data operasional memungkinkan.
- Tambahkan data time series yang lebih lengkap untuk memonitor perubahan Cancellation Rate dari waktu ke waktu.
- Validasi business threshold untuk long-lead dan potential room value bersama Revenue/Reservation Manager.
